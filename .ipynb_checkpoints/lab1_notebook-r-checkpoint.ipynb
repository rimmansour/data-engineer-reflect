## <a name="ex1">Exercise 1: Ehrenfest model</a> [(&#8593;)](#content) <!-- [$\cdot$/10] -->
 Consider a system of $K = 30$ particles (labeled from 1 to $K$) evolving in a closed box. The box is divided into two compartments in contact with each other, respectively identified by an index, $0$ and $1$. A hole at the interface between the two compartments allows the particles to move from one compartment to the other.

 Particle motion is modeled as follows: at each discrete time intant $n$, one particle is chosen uniformly at random and moved from its current compartment to the other. Let $X(n)$ denote the number of particles in compartment $0$ at time $n$.

3\. <!--[$\cdot$/0.5]--> Recall the structure of the transition matrix, and encode it in Python (without any for loop).

Hint: use the function `numpy.diag`.

>$P$ est une matrice de taille $(K+1) \cdot (K+1)$ tel que :  
>$P_{0,1}=P_{K,K-1}=1$  
>Pour tout $i \in$ {$1, ..., K-1$}, $P_{i,i+1}= 1-i/K$ et $P_{i,i-1}= i/K$

````python
K=30
P = np.diag(np.arange(1,0, -1/K),1)+np.diag(np.arange(1,0, -1/K)[::-1],-1)
````

5\. <!--[$\cdot$/2]--> Implement a Python function `ehrenfest` to simulate a trajectory of the chain for a system of $K$ particles, for initial distribution $\mu$ (for instance a Dirac centered in $0$, meaning that the compartment $0$ is empty at $n = 0$). The maximum number of time steps will be controlled by an input parameter $n_{\max}$. Random number generation will be controlled by a [random number generator](https://numpy.org/doc/stable/reference/random/generator.html) passed as an input to the function.

For an efficient implementation, **do not use vector-matrix product with the transition matrix implemented in 3**: rely on the description of the system instead.


````python
def ehrenfest(K : int, mu : np.ndarray, rng = np.random.default_rng(12345), nmax = 5000) -> np.ndarray :
    """
    Return the trajectory of the chain for a system of K particles for initial distribution mu.
    The maximum number of time steps is controlled by the input parameter nmax, preventing the algorithm from running too long.
    The random number generation is controlled by the random number generator parameter rng.

    Parameters
    ----------
    K : int
        The number of particles
    mu : np.ndarray
        The initial distribution
    rng : np.random._generator.Generator
        Random number generator to control the random number generation (default value is np.random.default_rng(12345))
    nmax : int
        The maximum number of time steps (default value is 5000)

    Returns
    -------
    np.ndarray 
        The trajectory of the chain of the system (Xn)
    """
    assert K>0, "The number of particles K must be strictly positive"
    assert len(mu) == K+1, "The initial distribution mu is not coherent with the number of particules K"
    assert sum(mu) == 1, "The initial distribution mu does not sum to 1"
    assert nmax>0, "The maximum number of time steps nmax must be strictly positive"
    T = np.empty(nmax)
    X0 = np.argmax(mu)
    T[0] = X0
    num_part = np.full(K, False)
    num_part[rng.choice(np.arange(K), X0,replace=False)]=True #Les numéros des X0 particules dans le compartiment 0
    for n in range(nmax) :
        particule = rng.integers(1,K,endpoint=True)
        if num_part[particule-1]==False:
            T[n] = T[n-1] + 1
            num_part[particule-1]=True
        else :
            T[n] = T[n-1] - 1
            num_part[particule-1]=False
    return T
````

8\. a) Modify the function defined in 5. so that it returns the return time to state 0, defined as $T_{0,0} = \inf \bigl\{ n > 0, X(n) = 0 \mid X(0) = 0 \bigr\}$.

````python
def ehrenfest_v2(K : int, mu : np.ndarray, rng = np.random.default_rng(12345), nmax = 5000) -> tuple :
    """
    Return the return time to the initial state of the chain for a system K particles for initial distribution mu.
    If the algorithm doesn't reach the initial state back, it returns nmax+3
    The maximum number of time steps is controlled by the input parameter nmax, preventing the algorithm from running too long.
    The random number generation is controlled by the random number generator parameter rng.

    Parameters
    ----------
    K : int
        The number of particles
    mu : np.ndarray
        The initial distribution (default value is 0)
    rng : np.random._generator.Generator
        Random number generator to control the random number generation (default value is np.random.default_rng(12345))
    nmax : int
        The maximum number of time steps (default value is 5000)

    Returns
    -------
    int
        The (first) return time to the initial state
    """
    assert K>0, "The number of particles K must be strictly positive"
    assert len(mu) == K+1, "The initial distribution mu is not coherent with the number of particules K"
    assert sum(mu) == 1, "The initial distribution mu does not sum to 1"
    assert nmax>0, "The maximum number of time steps nmax must be strictly positive"
    T = np.empty(nmax)
    X0 = np.argmax(mu)
    T[0] = X0
    num_part = np.full(K, False)
    num_part[rng.choice(np.arange(K), X0,replace=False)]=True #Les numéros des X0 particules dans le compartiment 0
    for n in range(nmax) :
        particule = rng.integers(1,K,endpoint=True)
        if num_part[particule-1]==False:
            T[n] = T[n-1] + 1
            num_part[particule-1]=True
        else :
            T[n] = T[n-1] - 1
            num_part[particule-1]=False
        if T[n] == X0:
            return n
    ## Si on ne retourne pas après nmax itérations, on retourne nmax + 3 pour savoir qu'on a pas fait de retour pour éviter une boucle empiriquement infinie. Plus le nombre d'états est grand, plus le temps moyen sera long.
    return nmax + 3
````

````python
rng = np.random.default_rng(12345)
K=10
T_return = ehrenfest_v2(K, np.array([1]+[0]*K), rng ,5000)
print(T_return)
````

8\. b) [**Optional**] Run several chains (about 5, ideally in parallel) for $K = 10$, $n_{\max} = 5000$, and compare the empirical average of $T_{0,0}$ to $\pi(0)$. What do you observe?
 > Hint: a good tutorial showing how to run functions in parallel in Python is available [here](https://www.machinelearningplus.com/python/parallel-processing-python/).

````python
N_simulations = 2000
nmax = 5000
T = np.zeros(N_simulations)
for i in range(N_simulations):
    seed = np.random.randint(1,9999)
    _, T_re = ehrenfest_v2(K = 10, X0 = 0, rng = np.random.default_rng(seed), nmax = nmax)
    T[i]=T_re
nb_nan = sum(np.isnan(T))
nb_notnan = N_simulations - nb_nan
print('Nombre de simulations où on atteint nmax avant de revenir à l\'état initial :',nb_nan)
print('Moyenne du temps de retour à l\'état initial sur les',nb_notnan,'autres simulations :',round(np.nanmean(T)))
````

>Pour être plus rigoureux dans le calcul de la moyenne du temps de retour à l'état initial, pour les simulations où on se retrouve avec une valeur None, le temps de retour est réellement supérieur à nmax. On peut alors calculer la moyenne de la manière suivante :

````python
T_moy = (np.nanmean(T)*nb_notnan+nmax*nb_nan)/N_simulations
T_moy
````

````python
pi0=ss.binom.pmf(0,10, .5)
pi0
````

````python
np.isclose(1/T_moy,pi0,atol=1e-04)
````

>On en déduit que la moyenne de $T_{0,0}$ tend vers $\frac {1}{\pi(0)}$

## <a name="ex2">Exercise 2: Simulation of a discrete time homogeneous Markov chain</a> [(&#8593;)](#content)
 Let $\bigl( X(n) \bigr)_{n \in \mathbb{N}}$ be a discrete time homogeneous Markov chain defined by the following initial distribution $\mu \in \mathbb{R}^K$ and transition matrix $P \in \mathbb{R}^{K \times K}$, with $K = 3$:

$$
  \mu = [0, 1, 0], 
  %
  \quad
  %
  P = \begin{pmatrix}
   0.2 & 0.7 & 0.1 \\
   0.9 & 0 & 0.1 \\
   0.2 & 0.8 & 0 
  \end{pmatrix}.
$$

1\. What can you say about the Markov chain $X$? (irreducibility, positive recurrence, periodicity, ...). Justify each of your claim, citing the relevant results from the lecture.

>Voici le diagramme d'état de cette chaîne de markov :

2\. Write a function `simulate_dthmc` simulating the trajectory of the Markov chain $\bigl( X(n) \bigr)_{n \in \mathbb{N}}$ for $n_{\max}$ time steps. The signature of the function should include the following elements:
   - list of inputs: transition matrix $P$, initial distribution $\mu$, number of time steps $n_{\max}$, [random number generator](https://numpy.org/doc/stable/reference/random/generator.html);
   - output: array of lenght $n_{\max}$ representing the trajectory.
   
**To this end, do not use matrix vector products, which would lead to an extremely inefficient algorithm in this case.**
   

````python
def simulate_dthmc(P : np.ndarray, mu : np.ndarray, rng = np.random.default_rng(12345), nmax = 5000) -> np.ndarray :
    """
    Return the trajectory of the chain for a system defined by the initial distribution mu and the transition matrix P. 
    The maximum number of time steps is controlled by the input parameter nmax.
    The random number generation is controlled by the random number generator parameter rng.

    Parameters
    ----------
    P : np.ndarray
        The transition matrix
    mu : np.ndarray
        The initial distribution
    rng : np.random._generator.Generator
        Random number generator to control the random number generation (default value is np.random.default_rng(12345))
    nmax : int
        The maximum number of time steps (default value is 5000)

    Returns
    -------
    np.ndarray 
        The trajectory of the chain of the system (Xn)
    """
    assert P.shape[0] == P.shape[1], "The transition matrix P must be a square matrix"
    K = P.shape[0]
    assert np.all(P>=0), "The transition matrix P must be a positif matrix"
    for i in range(K):
        assert np.isclose(sum(P[i,:]) , 1), "The transition matrix P must be a stochastic matrix"
    assert  len(mu) == K, "The initial distribution mu is not coherent with the shape of the transition matrix P"
    assert nmax>0, "The maximum number of time steps nmax must be strictly positive"
    T = np.empty(nmax)
    T[0] = np.argmax(mu)
    for n in range(1,nmax):
        T[n] = rng.choice([0,1,2], p=P[int(T[n-1]),:])
    return T
````

3\. Simulate a trajectory of the chain for $n_{\max} = 2000$ starting from $X(0) = 1$. Plot the histogram of the states visited by the chain.
 

````python
P = np.array([
        [0.2,0.7,0.1],
        [0.9,0,0.1],
        [0.2,0.8,0]])
distribution_init = [0,1,0]
Nmax = 2000
````

````python
Trajectory = simulate_dthmc(P, distribution_init, np.random.default_rng(12345), Nmax)
h= np.array([sum(Trajectory==0),sum(Trajectory==1),sum(Trajectory==2)])
plt.bar([0,1,2],height=h/Nmax)
plt.title('Histogram of the trajectory obtained')
plt.show()
````

4\. Determine numerically an invariant distribution $\boldsymbol{\pi}$ of the chain (*e.g.*, based on an eigendecomposition [numpy.linalg.eig](https://docs.scipy.org/doc/numpy/reference/generated/numpy.linalg.eig.html)). Is it unique? Compare it to the histogram obtained in 2 (superimpose graphs). What can you conclude?
 

````python
eigenvalues, eigenvectors = np.linalg.eig(P.T)
````

````python
pi = eigenvectors[:,np.isclose(eigenvalues,1)].reshape(3)
pi_normalisé = pi/sum(pi)
pi_normalisé
````

````python
np.allclose(pi_normalisé , pi_normalisé@P)
````

````python
sum(pi_normalisé)
````

````python
plt.bar([0,1,2],height=h/Nmax, label='Trajectory histogram')
plt.plot(pi_normalisé,label='Invariant distribution $\pi$',color='r')
plt.title('Histogram of the trajectory obtained Vs theoretical limit distribution $\pi$')
plt.legend()
plt.show()
````

>On obtient ainsi un seul vecteur $\boldsymbol{\pi}$ positif qui vérifie $\boldsymbol{\pi}=\boldsymbol{\pi} \cdot P$ et $ \sum_{j} {\boldsymbol{\pi}}_{j} = 1 $ donc c'est une probabilité invariante et il est unique

5\. a) Compute $\mu_n \triangleq \mu P^n$, the probability distribution of $X(n)$. What is the limit of $\mu_n$ as $n$ goes to $+\infty$? Illustrate the result numerically.


````python
def MuN(N):
    P_n = np.array([np.linalg.matrix_power(P,n) for n in range(N)])
    mu_n = np.array([distribution_init @ P_n[n] for n in range(N)])
    return mu_n
````

````python
N=50
mu_n = MuN(N)
np.allclose(pi_normalisé ,mu_n[-1])
````

6\. For each state $i \in \{0, \dotsc, K-1 \}$, simulate 100 trajectories starting from the state $i$ until the return time to $i$. For each state, compute the (empirical) average return time. Compare with its theoretical value.
 

````python
def simulate_dthmc_v2(P : np.ndarray, mu : np.ndarray, rng = np.random.default_rng(12345), nmax = 5000) -> tuple :
    """
    Return the trajectory of the chain as well as the return time to the initial state for a system defined by the initial distribution mu and the transition matrix P. 
    The maximum number of time steps is controlled by the input parameter nmax.
    The random number generation is controlled by the random number generator parameter rng.

    Parameters
    ----------
    P : np.ndarray
        The transition matrix
    mu : np.ndarray
        The initial distribution
    rng : np.random._generator.Generator
        Random number generator to control the random number generation (default value is np.random.default_rng(12345))
    nmax : int
        The maximum number of time steps (default value is 5000)

    Returns
    -------
    (np.ndarray, int)
        The trajectory of the chain of the system (Xn)
        The (first) return time to the initial state X0
    """
    assert P.shape[0] == P.shape[1], "The transition matrix P must be a square matrix"
    K = P.shape[0]
    assert np.all(P>=0), "The transition matrix P must be a positif matrix"
    for i in range(K):
        assert np.isclose(sum(P[i,:]) , 1), "The transition matrix P must be a stochastic matrix"
    assert  len(mu) == K, "The initial distribution mu is not coherent with the shape of the transition matrix P"
    assert nmax>0, "The maximum number of time steps nmax must be strictly positive"
    T = np.empty(nmax)
    n_return = []
    T[0] = np.argmax(mu)
    for n in range(1,nmax):
        T[n] = rng.choice([0,1,2], p=P[int(T[n-1]),:])
        if T[n] == np.argmax(mu):
            n_return.append(n)
    if n_return == []:
        n_r = None
    else :
        n_r = n_return[0]
    return T, n_r
````

````python
N_simulations = 100
for i in [0,1,2]:
    print(f'--- i = {i} ---')
    distribution_init = [0,0,0]
    distribution_init[i]=1
    T = np.zeros(N_simulations)
    for j in range(N_simulations):
        seed = np.random.randint(1,9999)
        _, T_re = simulate_dthmc_v2(P, distribution_init, np.random.default_rng(seed), Nmax)
        T[j] = T_re
    print(f'\tMoyenne du temps de retour à l\'état initial sur les {N_simulations} simulations :',np.mean(T))
````
